# Online Shopper Purchase Intention Prediction Using ANN

## 1. Project Overview

The aim of this project is to develop an **Artificial Neural Network (ANN) binary classification model** to predict whether an online shopping session will result in a purchase.

The model uses information about a visitor's browsing behaviour and session characteristics, such as the number of pages visited, time spent on different types of pages, bounce rate, exit rate, page value, visitor type, traffic source, and other session-related information.

The target variable is **`Revenue`**:

- `False` → No Purchase
- `True` → Purchase

Therefore, this project is a **supervised binary classification problem**.

---

## 2. Business Problem

Online shopping websites receive many visitors, but only a small proportion of browsing sessions result in purchases.

Predicting the purchase intention of visitors can help an e-commerce business:

- Identify visitors who are more likely to purchase
- Understand customer browsing behaviour
- Improve marketing strategies
- Personalize customer experiences
- Improve conversion rates
- Make better data-driven business decisions

The main question addressed by this project is:

> **Can the browsing behaviour and session characteristics of an online visitor be used to predict whether the session will result in a purchase?**

---

## 3. Project Objective

The main objective is to build an **Artificial Neural Network classification model** that predicts whether an online shopping session will generate revenue.

The model will learn patterns from historical shopping-session data and estimate the probability that a new session will result in a purchase.

## 6. Dataset Columns

The dataset contains **18 columns**, including 17 input features and one target variable.

| Column | Type | Detailed Description |
|---|---|---|
| **Administrative** | Numerical | Number of administrative pages visited by the user during the shopping session. These may include pages associated with account or administrative activities. |
| **Administrative_Duration** | Numerical | Total amount of time the visitor spent on administrative pages during the session. |
| **Informational** | Numerical | Number of informational pages visited during the session. These pages generally provide information rather than displaying products. |
| **Informational_Duration** | Numerical | Total amount of time the visitor spent on informational pages during the session. |
| **ProductRelated** | Numerical | Number of product-related pages visited during the shopping session. It represents the visitor's level of interaction with product pages. |
| **ProductRelated_Duration** | Numerical | Total amount of time spent viewing product-related pages during the session. |
| **BounceRates** | Numerical | Average bounce rate associated with the pages visited during the session. A bounce generally occurs when a visitor enters a page and leaves without further interaction. Higher values may indicate lower engagement. |
| **ExitRates** | Numerical | Average exit rate associated with the pages visited during the session. It indicates how frequently those pages are the final pages viewed before leaving the website. |
| **PageValues** | Numerical | Average value of the pages visited before completing an e-commerce transaction. It provides information about how strongly the visited pages are associated with successful purchases. |
| **SpecialDay** | Numerical | Indicates how close the shopping session occurred to a special day, such as Valentine's Day or Mother's Day. Higher values indicate greater proximity to a special occasion. |
| **Month** | Categorical | Month in which the online shopping session occurred. |
| **OperatingSystems** | Encoded Categorical | Integer category code representing the operating system used by the visitor. The numbers identify different operating-system categories and are not continuous numerical measurements. |
| **Browser** | Encoded Categorical | Integer category code representing the browser used during the shopping session. The numbers represent different browser categories rather than numerical quantities. |
| **Region** | Encoded Categorical | Integer category code representing the geographical region associated with the visitor. The values identify different regions and do not represent numerical magnitude. |
| **TrafficType** | Encoded Categorical | Integer category code representing the traffic source or channel through which the visitor reached the website. |
| **VisitorType** | Categorical | Type of visitor using the website. Categories include `Returning_Visitor`, `New_Visitor`, and `Other`. |
| **Weekend** | Boolean | Indicates whether the shopping session occurred during the weekend. `True` represents a weekend session and `False` represents a non-weekend session. |
| **Revenue** | Boolean / Target | Target variable indicating whether the shopping session resulted in a purchase. `True` represents Purchase and `False` represents No Purchase. |

## 7. Target Variable

The target variable for the project is **`Revenue`**.

It contains two classes:

- `False` → No Purchase
- `True` → Purchase

For ANN model building, these classes can be represented as:

- `0` → No Purchase
- `1` → Purchase

Since there are only two possible outcomes, this is a **binary classification problem**.

## 10. Machine Learning Problem

The objective of the model is to answer:

> **Will an online shopping session result in a purchase?**

The ANN will use visitor behaviour and session information as input and predict one of two outcomes:

- **No Purchase**
- **Purchase**

Therefore, this is a **binary classification problem**.

---

## 11. ANN Architecture

The basic ANN structure for this project will be:

**Input Features**

↓

**Hidden Layer 1 — ReLU**

↓

**Hidden Layer 2 — ReLU**

↓

**Output Layer — 1 Neuron + Sigmoid**

↓

**Probability of Purchase**

↓

**Purchase / No Purchase**

Since this is binary classification, the output layer uses **one neuron with a Sigmoid activation function**.

The sigmoid output represents the estimated probability of the positive class:

**`Revenue = 1` → Purchase**

The exact ANN architecture and hyperparameters will be determined during model building.

---

## 12. Model Evaluation

Since the target variable is imbalanced, multiple classification metrics will be considered.

- **Accuracy** → Overall proportion of correct predictions.
- **Precision** → Of the sessions predicted as purchases, how many actually resulted in purchases.
- **Recall** → Of the actual purchasing sessions, how many were correctly identified.
- **F1-Score** → Provides a balance between Precision and Recall.
- **ROC-AUC** → Measures the model's ability to distinguish between Purchase and No Purchase across different thresholds.
- **Confusion Matrix** → Shows True Positives, True Negatives, False Positives, and False Negatives.

These correlations represent **associations rather than causation**. Features with weak linear correlations may still provide useful information to the ANN because neural networks can learn nonlinear relationships between variables.
